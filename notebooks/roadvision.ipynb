```python
import kagglehub

dataset_path = kagglehub.dataset_download(
    "ashfakyeafi/road-vehicle-images-dataset"
)

print("Dataset  :", dataset_path)
```

```python
from pathlib import Path
import shutil

# Dossier racine du projet
project_path = Path("/content/detection_des_vehicule")

# Destination du dataset
destination = project_path / "data" / "raw"
destination.mkdir(parents=True, exist_ok=True)

# Copier les fichiers du dataset
for item in Path(dataset_path).iterdir():
    target = destination / item.name

    if item.is_dir():
        shutil.copytree(item, target, dirs_exist_ok=True)
    else:
        shutil.copy2(item, target)

print("Dataset copie dans :", destination)
print("\nContenu :")

for item in sorted(destination.iterdir()):
    print("[Dossier]" if item.is_dir() else "[Fichier]", item.name)
```
